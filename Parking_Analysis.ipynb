```python
junction_summary = (
    df[df["junction_name"] != "No Junction"]
    .groupby("junction_name")
    .agg(
        violations=("id","count"),
        weekend_rate=("is_weekend","mean"),
        peak_hour_rate=("peak_hour","mean")
    )
    .reset_index()
)

junction_summary.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

junction_summary[
    ["violations_norm",
     "weekend_norm",
     "peak_norm"]
] = scaler.fit_transform(
    junction_summary[
        ["violations",
         "weekend_rate",
         "peak_hour_rate"]
    ]
)
```

```python
junction_summary["risk_score"] = (
    junction_summary[
        ["violations_norm",
         "weekend_norm",
         "peak_norm"]
    ].mean(axis=1)
) * 100
```

```python
def classify(score):
    if score >= 70:
        return "High"
    elif score >= 40:
        return "Medium"
    else:
        return "Low"

junction_summary["risk_level"] = junction_summary[
    "risk_score"
].apply(classify)
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertia = []

K = range(2, 15)

for k in K:
    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    km.fit(coords[["latitude", "longitude"]])
    inertia.append(km.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K, inertia, marker='o')
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.grid(True)

plt.show()
```

```python
daily_data = (
    df[df["junction_name"] != "No Junction"]
    .groupby([
        df["created_datetime"].dt.date,
        "junction_name"
    ])
    .size()
    .reset_index(name="violations")
)

daily_data.head()
```

```python
daily_data.shape
```

```python
top_junctions = (
    df[df["junction_name"] != "No Junction"]
    .groupby("junction_name")
    .size()
    .sort_values(ascending=False)
    .head(20)
    .index
)

top_junctions
```

```python
daily_data["date"] = pd.to_datetime(daily_data["created_datetime"])

latest_date = daily_data["date"].max()

future_predictions = []

for junction in top_junctions:

    temp = daily_data[
        daily_data["junction_name"] == junction
    ].copy()

    recent_week = temp[
        temp["date"] > latest_date - pd.Timedelta(days=7)
    ]["violations"].mean()

    previous_week = temp[
        (temp["date"] > latest_date - pd.Timedelta(days=14))
        &
        (temp["date"] <= latest_date - pd.Timedelta(days=7))
    ]["violations"].mean()

    growth = recent_week - previous_week

    predicted_next_week = recent_week + growth

    future_predictions.append([
        junction,
        round(recent_week, 2),
        round(predicted_next_week, 2)
    ])
```

```python
future_hotspots = pd.DataFrame(
    future_predictions,
    columns=[
        "junction_name",
        "current_week_avg",
        "predicted_next_week_avg"
    ]
)

future_hotspots.sort_values(
    "predicted_next_week_avg",
    ascending=False
).head(20)
```

```python
future_hotspots["growth_percent"] = (
    (
        future_hotspots["predicted_next_week_avg"]
        -
        future_hotspots["current_week_avg"]
    )
    /
    future_hotspots["current_week_avg"]
) * 100

future_hotspots.sort_values(
    "growth_percent",
    ascending=False
)
```

```python
# Get top 20 junctions

top_junctions = (
    df[df["junction_name"] != "No Junction"]
    .groupby("junction_name")
    .size()
    .sort_values(ascending=False)
    .head(20)
    .index
)

# Latest date in dataset

latest_date = daily_data["date"].max()

future_predictions = []

for junction in top_junctions:

    temp = daily_data[
        daily_data["junction_name"] == junction
    ].copy()

    recent_week = temp[
        temp["date"] > latest_date - pd.Timedelta(days=7)
    ]["violations"].mean()

    previous_week = temp[
        (temp["date"] > latest_date - pd.Timedelta(days=14))
        &
        (temp["date"] <= latest_date - pd.Timedelta(days=7))
    ]["violations"].mean()

    growth = recent_week - previous_week

    # Prevent negative predictions

    predicted_next_week = max(
        recent_week + growth,
        0
    )

    future_predictions.append([
        junction,
        round(recent_week, 2),
        round(predicted_next_week, 2)
    ])

# Create dataframe

future_hotspots = pd.DataFrame(
    future_predictions,
    columns=[
        "junction_name",
        "current_week_avg",
        "predicted_next_week_avg"
    ]
)

# Growth percentage

future_hotspots["growth_percent"] = (
    (
        future_hotspots["predicted_next_week_avg"]
        -
        future_hotspots["current_week_avg"]
    )
    /
    future_hotspots["current_week_avg"]
) * 100

# Trend classification

future_hotspots["trend"] = future_hotspots[
    "growth_percent"
].apply(
    lambda x:
    "Increasing"
    if x > 15
    else (
        "Decreasing"
        if x < -15
        else "Stable"
    )
)

# View results

future_hotspots.sort_values(
    "growth_percent",
    ascending=False
)
```
